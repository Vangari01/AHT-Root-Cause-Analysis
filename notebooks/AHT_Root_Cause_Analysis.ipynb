```python
#Comparing average AHT by Repeated Messages
```

```python
#Comparing average AHT by Navigation time 
```

```python
df["Navigation_Group"]=pd.cut(df["Navigation_Time_sec"], 
bins=[0, 25, 30, 35, 40, 100], labels=["0-25", "26-30","31-35","36-40", "40+"])
```

```python
df.groupby("Navigation_Group")["AHT_sec"].mean().sort_values()
```

```python
#Comparing average AHT by Customer Response Time 
```

```python
df["Customer_Responsetime_Group"]=pd.cut(df["Customer_Response_sec"], 
bins=[40, 50, 60, 70, 80, 90], labels=["40-50", "51-60","61-70","71-80", "81+"])
```

```python
df.groupby("Customer_Responsetime_Group")["AHT_sec"].mean().sort_values()
```

```python
#Comparing AHT by Agent Response Time 
```

```python
df["Agent_Responsetime_Group"]=pd.cut(df["Agent_Response_sec"], 
bins=[20, 30, 40, 50, 60,70], labels=["20-30", "31-40","41-50","51-60", "61+"])
```

```python
df.groupby("Agent_Responsetime_Group")["AHT_sec"].mean().sort_values()
```
